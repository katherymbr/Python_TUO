# Python TUO - Solucion propuesta para el ejercicio integrador: Gestión de Inventario del centro oceanográfico OceanLab

## Objetivo
Vamos a desarrollar un programa en Python para gestionar los equipos e insumos de OceanLab. El inventario es una `lista de tuplas`. Cada tupla contiene: el nombre, el código de identificación, la cantidad disponible y el precio unitario.

El programa presentará un menú que se repetirá mediante un ciclo `while` hasta que el usuario seleccione la opción 6.

**Operaciones del menú**
1. Agregar un equipo o insumo.
2. Eliminar un equipo o insumo.
3. Actualizar la cantidad de un equipo o insumo.
4. Calcular el valor total del inventario.
5. Mostrar el inventario.
6. Salir.

No se utilizan funciones definidas por el usuario ni `break`. Solo lo que se ha visto en el curso!


## Menú de opciones y finalización

Utilizaremos un ciclo `while` para que el menú se muestre repetidamente hasta que el usuario seleccione la opción 6. La variable `opcion` almacena la elección. La estructura `if`, `elif` y `else` permite ejecutar la operación correspondiente. Si el usuario selecciona 6, se muestra un mensaje y el ciclo termina cuando se evalúa nuevamente su condición. Si ingresa una opción que no corresponde, se informa que no es válida.

## Resumen de funciones y métodos utilizados

- `print()`: muestra información en pantalla.
- `input()`: recibe los datos que escribe el usuario; devuelve texto.
- `int()`: convierte un dato en un número entero.
- `float()`: convierte un dato en un número decimal.
- `len(inventario)`: devuelve la cantidad de elementos de la lista.
- `append()`: agrega un elemento al final de la lista.
- `pop(i)`: elimina el elemento ubicado en la posición `i`.
- `round(total, 2)`: redondea el resultado a dos decimales.

**Importante:** `while`, `for`, `if`, `elif` y `else` son estructuras de control, no funciones. El código supone que el usuario ingresa números válidos para la opción, la cantidad y el precio.


## a) Agregar un equipo o insumo

Vamos a comprobar si el inventario contiene menos de 10 elementos. Si hay espacio disponible, solicitaremos el nombre, el código, la cantidad y el precio unitario del nuevo equipo o insumo. Con esos datos crearemos una tupla y la agregaremos al final de la lista mediante `append()`. Si el inventario ya contiene 10 elementos, informaremos que no es posible agregar más.


## b) Eliminar un equipo o insumo

Vamos a solicitar el código del elemento que se desea eliminar. Recorreremos el inventario con un ciclo `while`, utilizando `i` para representar la posición actual y `encontrado` para indicar si encontramos el código. En cada posición compararemos el código ingresado con el segundo elemento de la tupla. Si coinciden, eliminaremos toda la tupla mediante `pop()` y marcaremos que fue encontrada. Si no coinciden, avanzaremos a la siguiente posición. Si terminamos la búsqueda sin encontrar el código, mostraremos un mensaje. La condición del ciclo detiene la búsqueda cuando se encuentra el elemento o se llega al final de la lista.


## c) Actualizar la cantidad de un equipo o insumo

Vamos a solicitar el código del elemento y la nueva cantidad disponible. Después, recorreremos el inventario con un ciclo `while` hasta encontrar el código o llegar al final de la lista. Cuando lo encontremos, recuperaremos el nombre y el precio unitario. Como las tuplas no se pueden modificar directamente, crearemos una nueva tupla con la cantidad actualizada y reemplazaremos la anterior en la lista. Si el código no se encuentra, informaremos que no existe.


## d) Calcular el valor total del inventario

Vamos a calcular el valor económico de todos los equipos e insumos registrados. Crearemos una variable `total`, inicializada en cero, que funcionará como acumulador. Mediante un ciclo `for`, recorreremos cada tupla, obtendremos la cantidad y el precio unitario, multiplicaremos ambos valores y sumaremos el resultado al acumulador. Al finalizar, mostraremos el valor total del inventario en dólares.


## e) Mostrar el inventario

Vamos a recorrer todos los elementos de la lista mediante un ciclo `for`. En cada iteración accederemos a los cuatro datos de la tupla mediante sus índices y mostraremos el nombre, el código de identificación, la cantidad disponible y el precio unitario. Así, el usuario podrá consultar todos los elementos registrados en el inventario.


## Programa en Python

```python
# Inventario inicial: es una lista de tuplas. Cada elemento es una tupla que tiene: nombre, código, cantidad y precio unitario.

inventario = [
    ("Termómetro", "TM100", 10, 28.50),
    ("Pipeta", "PP50", 5, 12.80),
    ("Probeta", "PB103", 8, 15.02)
]

# La variable opcion almacena la elección del usuario.
#  Empezamos con opcion = 0 para que el menú se muestre por primera vez
opcion = 0

# El menú se repite mientras la opción sea distinta de 6.
while opcion != 6:
    print("\n----- MENÚ DE INVENTARIO -----")
    print("1. Agregar un equipo o insumo")
    print("2. Eliminar un equipo o insumo")
    print("3. Actualizar la cantidad de un equipo o insumo")
    print("4. Calcular el valor total del inventario")
    print("5. Mostrar el inventario")
    print("6. Salir")

    # input() recibe texto; int() convierte la opción en un número entero.
    opcion = int(input("Seleccione una opción: "))

    #*********************************************************************************
    # a) AGREGAR UN EQUIPO O INSUMO
    #*********************************************************************************

    # Comprobamos que el inventario tenga menos de 10 elementos.
    if opcion == 1:
        if len(inventario) < 10:
            # Solicitamos al usuario los datos del nuevo elemento.
            nombre = input("Ingrese el nombre: ")
            codigo = input("Ingrese el código: ")
            cantidad = int(input("Ingrese la cantidad disponible: "))
            precio = float(input("Ingrese el precio unitario en dólares: "))

            # Creamos una tupla con los cuatro datos ingresados.
            nuevo_componente = (nombre, codigo, cantidad, precio)

            # append() agrega la nueva tupla al final de la lista llamada inventario.
            inventario.append(nuevo_componente)
            print("Equipo o insumo agregado correctamente.")
        else:
            # Si ya hay 10 elementos, no permitimos agregar otro.
            print("No es posible agregar más elementos.")
            print("El inventario ya tiene 10 componentes.")

    #*********************************************************************************
    # b) ELIMINAR UN EQUIPO O INSUMO
    #*********************************************************************************

    elif opcion == 2:
        # Solicitamos el código del elemento que se desea eliminar. Debe ser un dato tipo "str"
        codigo = input("Ingrese el código del elemento que desea eliminar: ")

        # encontrado indica si hallamos el código solicitado.
        encontrado = False

        # Comenzamos la búsqueda en la primera posición de la lista.
        i = 0

        # Buscamos mientras i esté dentro de la lista y no se haya encontrado el código.
        while i < len(inventario) and encontrado == False:
            # inventario[i] selecciona una tupla y [1] accede a su código.
            if inventario[i][1] == codigo:
                # El métdodo pop(i) elimina de la lista la tupla ubicada en la posición i.
                inventario.pop(i)
                encontrado = True
                # No incrementamos i porque el siguiente elemento ocupa esta posición.
            else:
                # Si el código no coincide, avanzamos a la siguiente posición.
                i = i + 1

        # Informamos si se eliminó el elemento o si el código no existe.
        if encontrado:
            print("Elemento eliminado correctamente.")
        else:
            print("El código no existe.")

    #*********************************************************************************
    # c) ACTUALIZAR LA CANTIDAD DE UN EQUIPO O INSUMO
    #*********************************************************************************

    elif opcion == 3:
        # Solicitamos el código y la nueva cantidad.
        codigo = input("Ingrese el código del elemento que desea actualizar: ")
        nueva_cantidad = int(input("Ingrese la nueva cantidad disponible: "))

        encontrado = False
        i = 0

        # Buscamos el elemento por su código.
        while i < len(inventario) and encontrado == False:
            if inventario[i][1] == codigo:
                # Recuperamos el nombre y el precio de la tupla existente.
                nombre = inventario[i][0]
                precio = inventario[i][3]

                # Las tuplas son inmutables: creamos una nueva con la cantidad actualizada.
                nuevo_componente = (nombre, codigo, nueva_cantidad, precio)

                # Reemplazamos la tupla anterior en la lista por la nueva.
                inventario[i] = nuevo_componente
                encontrado = True
            else:
                # Si el código no coincide, revisamos la siguiente posición.
                i = i + 1

        if encontrado:
            print("Cantidad actualizada correctamente.")
        else:
            print("El código no existe.")

    #*********************************************************************************
    # d) CALCULAR EL VALOR TOTAL DEL INVENTARIO
    #*********************************************************************************

    elif opcion == 4:
        # total es un acumulador que comienza en cero.
        total = 0

        # Recorremos cada tupla de la lista.
        for componente in inventario:
        # En cada tupla, la posición [2] es la cantidad y la posición [3] es el precio unitario.
        # Vamos a recorrer toda la lista y, para cada tupla, guardar esos valores.
        # Para ello, creamos las variables cantidad y precio
            cantidad = componente[2]
            precio = componente[3]

            # Sumamos al acumulador el valor del elemento: cantidad por precio.
            total = total + cantidad * precio

        # round(total, 2) redondea el resultado a dos decimales.
        print("El valor total del inventario es: $", round(total, 2))

    #*********************************************************************************
    # e) MOSTRAR EL INVENTARIO
    #*********************************************************************************

    elif opcion == 5:
        print("\n----- INVENTARIO -----")

        # Recorremos todas las tuplas del inventario.
        for componente in inventario:
            # Mostramos cada dato usando su índice dentro de la tupla.
            print("Nombre:", componente[0])
            print("Código:", componente[1])
            print("Cantidad:", componente[2])
            print("Precio unitario: $", componente[3])
            print("----------------------")

    # Si se elige 6, el ciclo termina cuando se evalúe nuevamente su condición.
    elif opcion == 6:
        print("Programa finalizado.")

    # Si se ingresa una opción distinta de 1 a 6, informamos que no es válida.
    else:
        print("Opción no válida.")
```
